### What was the Longest Phone Call and Longest Data Entry?


```python
print("   Longest call entry:")
data1=data[(data[...,3]=="call") & (data[...,6]=='mobile')]
MAX=np.max(data1[...,2].astype('int'))
print(data1[data1[...,2]==str(MAX)])

print("   Longest data entry:")
data1=data[data[...,3]=="data"]
MAX=np.max(data1[...,2].astype('float'))
print(data1[data1[...,2]==str(MAX)])
```

### How many seconds of phone calls are recorded in total?



```python
data1=data[(data[...,3]=="call") & (data[...,6]=='mobile')]
s=np.sum(data1[...,2].astype('float'))
s
```

### How many entries are there for each month?


```python
a=collections.Counter(data[...,4])
print(a)
```

### Get the first entry for each month


```python
for x in np.unique(data[...,4]):
    d2=data[data[...,4]==x]
    print(d2[0])
```

### Get the sum of the durations per month

```python
for x in np.unique(data[...,4]):
    d2=data[data[...,4]==x]
    print(f"{x} : {sum(d2[...,2].astype('float'))}")
```

### What is the sum of durations, for calls only, to each network


```python
for x in np.unique(data[...,5]):
    d2=data[(data[...,5]==x) & (data[...,3]=="call")]
    print(f"{x} : {sum(d2[...,2].astype('float'))}")
```

### How many calls, sms, and data entries/Duration are in each month?


```python
for x in np.unique(data[...,4]):
    print(f"Entries for month {x}")
    d2=data[data[...,4]==x]
    a=collections.Counter(d2[...,3])
    print(a)

print()
for x in np.unique(data[...,4]):
    d2=data[(data[...,4]==x) & (data[...,3]=="call")]
    print(f"Duration of calls in {x} : {sum(d2[...,2].astype('float'))}")
    d2=data[(data[...,4]==x) & (data[...,3]=="sms")]
    print(f"Duration of sms in {x} : {sum(d2[...,2].astype('float'))}")
    d2=data[(data[...,4]==x) & (data[...,3]=="data")]
    print(f"Duration of data in {x} : {sum(d2[...,2].astype('float'))}")
```

### How many calls, texts, and data are sent per month, split by network?


```python
for x in np.unique(data[...,4]):
    print(f"\nEntries for month {x} :")
    for y in np.unique(data[...,5]):
        print(y,end=" ")
        d2=data[(data[...,4]==x) & (data[...,5]==y)]
        a=collections.Counter(d2[...,3])
        print(a)
```
