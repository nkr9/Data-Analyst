#### Q2. (d) Player with Maximum Runs


```python

runs_scored = ball_data[:, 2].astype(int)
striker_ids = ball_data[:, 6]

unique_players, total_runs = np.unique(striker_ids, return_counts=True)
total_runs_per_player = {player: 0 for player in unique_players}
for i, player in enumerate(striker_ids):
    total_runs_per_player[player] += runs_scored[i]

max_runs_player_id = max(total_runs_per_player, key=total_runs_per_player.get)
max_runs_player_name = player_data[player_data[:, 0] == max_runs_player_id, 1][0]

print(f"Player with maximum runs: {max_runs_player_name}")
print(f"Total runs: {total_runs_per_player[max_runs_player_id]}")
```

#### 2. (e) Player who faced maximum balls

```python

striker_ids = ball_data[:, 6]
unique_players, balls_faced = np.unique(striker_ids, return_counts=True)
max_balls_player_id = unique_players[np.argmax(balls_faced)]
max_balls_player_name = player_data[player_data[:, 0] == max_balls_player_id, 1][0]

print(f"Player who faced maximum balls: {max_balls_player_name}")
print(f"Number of balls faced: {np.max(balls_faced)}")
```
